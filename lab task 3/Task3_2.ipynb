```python
class Search:
    graph = {
        "A":["B","C"],
        "B":["D","E"],
        "C":["F"],
        "D":[],
        "E":["G"],
        "F":[],
        "G":[]
    }
    def dls(self, node, goal, limit, path):
        if node == goal:
            return path
        if limit ==0:
            return None
        for n in self.graph[node]:
            if n not in path:
                result = self.dls(n, goal, limit - 1, path + [n])
                if result:
                    return result
        return None
    def iddfs(self, start, goal, maximum):
        for limit in range(maximum + 1):
            result = self.dls(start, goal, limit, [start])
            if result:
                return result
s = Search()
for limit in range(1,5):
    print(s.dls("A", "G", limit, ["A"]))
print(s.iddfs("A","G",4))
```
