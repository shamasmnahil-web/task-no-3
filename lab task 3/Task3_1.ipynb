```python
from collections import deque
class Search:
    graph = {
        "A":["B","C"],
        "B":["D","E"],
        "C":["F"],
        "D":[],
        "E":["G"],
        "F":[],
        "G":[]
    }
    def bfs(self, start, goal):
        q = deque([[start]])
        while q:
            path = q.popleft()
            n = path[-1]
            if n == goal:
                return path
            for x in self.graph[n]:
                q.append(path + [x])
    def dfs(self, start, goal):
        stack = [[start]]
        while stack:
            path = stack.pop()
            n = path[-1]
            if n == goal:
                return path
            for x in reversed(self.graph[n]):
                stack.append(path + [x])
s = Search()
print("BFS:", s.bfs("A","G"))
print("DFS:", s.dfs("A","G"))
```
