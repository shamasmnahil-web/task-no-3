```python
import heapq

class Search:
    graph = {
        "A":[("B",1),("C",6)],
        "B":[("D",2),("E",1)],
        "C":[("F",2),("G",12)],
        "D":[], "E":[("G",1)], "F":[], "G":[],
    }
    def ucs(self, start, goal):
        q = [(0, start,[start])]
        best = {start:0}

        while q:
            cost, node, path = heapq.heappop(q)

            if cost != best[node]:
                continue
            if node == goal:
                return path, cost
            for n, w in self.graph[node]:
                new_cost = cost + w
                if new_cost < best.get(n, float("inf")):
                    best[n] = new_cost
                    heapq.heappush(q, (new_cost, n, path + [n]))
s = Search()
print(s.ucs("A", "G"))
```
